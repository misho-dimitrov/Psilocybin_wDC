```python
import pandas as pd
```

<span style="font-size:160%"><b>Baseline</b></span>

```python
# Load your CSV
df = pd.read_csv("voxel_table_with_clusters_baseline.csv")

# a) Voxels per AnatomicalRegion
voxels_per_region = df['AnatomicalRegion'].value_counts().reset_index()
voxels_per_region.columns = ['AnatomicalRegion', 'VoxelCount']

# Map each region to its most common YeoNetwork (mode)
region_to_network = df.groupby('AnatomicalRegion')['YeoNetwork'] \
                      .agg(lambda x: x.mode().iat[0] if not x.mode().empty else 'Unknown') \
                      .reset_index()
region_to_network.columns = ['AnatomicalRegion', 'MostCommonYeoNetwork']

# Merge voxel counts with network info
voxels_per_region = pd.merge(voxels_per_region, region_to_network, on='AnatomicalRegion')

print("a) Voxels per Anatomical Region:")
print(voxels_per_region)

# b) Voxels per Yeo network
voxels_per_network = df['YeoNetwork'].value_counts().reset_index()
voxels_per_network.columns = ['YeoNetwork', 'VoxelCount']
print("\nb) Voxels per Yeo Network:")
print(voxels_per_network)

# c) Ranking by max Z-score per region
max_z_by_region = df.groupby('AnatomicalRegion')['Zvalue'].max().reset_index()
max_z_by_region = pd.merge(max_z_by_region, region_to_network, on='AnatomicalRegion')
max_z_by_region = max_z_by_region.sort_values(by='Zvalue', ascending=False)

print("\nc) Regions with highest Z-scores:")
print(max_z_by_region)

# d) Ranking by max Z-score per network
max_z_by_network = df.groupby('YeoNetwork')['Zvalue'].max().reset_index()
max_z_by_network = max_z_by_network.sort_values(by='Zvalue', ascending=False)

print("\nd) Networks with highest Z-scores:")
print(max_z_by_network)

# === Regions ranking in top 10 for both voxel count and Z-score ===
region_summary = pd.merge(voxels_per_region, max_z_by_region, 
                          on=['AnatomicalRegion', 'MostCommonYeoNetwork'])

region_summary['VoxelRank'] = region_summary['VoxelCount'].rank(method='min', ascending=False)
region_summary['ZscoreRank'] = region_summary['Zvalue'].rank(method='min', ascending=False)

top_regions = region_summary[(region_summary['VoxelRank'] <= 10) & 
                             (region_summary['ZscoreRank'] <= 10)]

print("\nRegions ranking in top 10 for both voxel count and Z-score:")
print(top_regions.sort_values(by=['VoxelRank', 'ZscoreRank']))

# Export all summary tables to Excel
with pd.ExcelWriter("summary_table_baseline.xlsx") as writer:
    voxels_per_region.to_excel(writer, sheet_name="Voxels_per_Region", index=False)
    voxels_per_network.to_excel(writer, sheet_name="Voxels_per_Network", index=False)
    max_z_by_region.to_excel(writer, sheet_name="MaxZ_by_Region", index=False)
    max_z_by_network.to_excel(writer, sheet_name="MaxZ_by_Network", index=False)
    top_regions.to_excel(writer, sheet_name="TopRegions", index=False)
```
